```python
ess_cols = [
    'cntry',    # country code (e.g. 'DE', 'FR')
    'essround', # ESS round number (1-11, corresponding to 2002-2023)
    'imbgeco',  # "Is immigration good or bad for economy?" (0=bad, 10=good)
    'imueclt',  # "Does immigration enrich or undermine cultural life?" (0=undermine, 10=enrich)
    'imwbcnt',  # "Is immigration good or bad for country overall?" (0=bad, 10=good)
    'brncntr',  # Born in country? (1=yes, 2=no) — used to filter to native-born only
    'eisced',   # Education level ISCED scale (0-7)
    'eduyrs',   # Years of education
    'agea',     # Age
    'gndr',     # Gender (1=male, 2=female)
    'uempla',   # Currently unemployed? (1=yes, 0=no)
    'lrscale',  # Left-right political scale (0=far left, 10=far right)
    'stfeco',   # Satisfaction with economy (0=extremely dissatisfied, 10=extremely satisfied)
]

ess = pd.read_csv('ESSQSS20.csv', usecols=ess_cols)

print(f"ESS loaded shape: {ess.shape}")
print(f"Rows: {len(ess):,} respondents")
print(f"Columns: {len(ess.columns)}")
print(f"\nESS rounds covered: {sorted(ess['essround'].unique())}")
print(f"Countries covered ({ess['cntry'].nunique()} total): {sorted(ess['cntry'].unique())}")
print(f"\nMissing values per column (raw — ESS uses high numbers like 77/88/99 for missing, not NaN):")
print(ess.isna().sum())
print(f"\nValue counts for brncntr (born in country — 1=yes, 2=no):")
print(ess['brncntr'].value_counts())
print(f"\nSample of raw attitude variables (values >10 are refusal/don't know codes):")
print(ess[['imbgeco', 'imueclt', 'imwbcnt']].describe())
ess.head()
```

```python
policy = pd.read_csv('migration_policy.xls', sep='\t')

policy['Year'] = pd.to_numeric(policy['Year'], errors='coerce')

print(f"DEMIG loaded shape: {policy.shape}")
print(f"Rows: {len(policy):,} policy changes")
print(f"\nColumn names:")
print(policy.columns.tolist())
print(f"\nCountries covered ({policy['Country'].nunique()} total):")
print(sorted(policy['Country'].dropna().unique()))
print(f"\nYear range: {policy['Year'].min()} to {policy['Year'].max()}")
policy.head(10)
```

```python
print("Restrictiveness value counts")
print(policy['Restrictiveness'].value_counts())

print("\nPolicy area value counts")
print(policy['Policy area'].value_counts())

print("\nMagnitude value counts")
print(policy['Magnitude'].value_counts())

print("\nMissing values per column")
print(policy.isna().sum())

print("\nSample of recent policy changes")
print(policy[['Country', 'Year', 'Summary', 'Restrictiveness', 'Magnitude', 'Policy area']].tail(10).to_string())
```
